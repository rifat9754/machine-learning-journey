# Multiple Linear Regression - House Price Prediction

## Project Objective

The objective of this project is to predict house prices
using multiple independent features through Multiple Linear Regression.

## Features

- Area
- Bedrooms
- Bathrooms
- House Age
- Location Score

## Target

- House Price

## Algorithms

1. Multiple Linear Regression from Scratch using NumPy
2. Multiple Linear Regression using Scikit-learn

## Evaluation Metrics

- MAE
- MSE
- RMSE
- R2 Score

# 1 - Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import warnings
warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

# 2 - Generate Dataset

```python
np.random.seed(42)

n = 500

# Generate independent features
area = np.random.randint(600, 3001, n)

bedrooms = np.clip(
    np.round(area / 550 + np.random.normal(0, 0.6, n)),
    1, 5
).astype(int)

bathrooms = np.clip(
    bedrooms - np.random.randint(0, 2, n),
    1, 4
)

age = np.random.randint(0, 41, n)

location_score = np.random.randint(1, 11, n)

# Generate random noise
noise = np.random.normal(0, 5, n)

# Generate house prices (in lakh BDT)
price = (
    15
    + (area * 0.035)
    + (bedrooms * 3)
    + (bathrooms * 2.5)
    - (age * 0.4)
    + (location_score * 4)
    + noise
)

# Create DataFrame
df = pd.DataFrame({
    "Area": area,
    "Bedrooms": bedrooms,
    "Bathrooms": bathrooms,
    "Age": age,
    "Location_Score": location_score,
    "Price": price
})

df.head()
```

### 2.1 Save and Load Dataset

```python
df.to_csv("data/house_prices.csv", index=False)

print("Dataset saved successfully!")

df = pd.read_csv("data/house_prices.csv")
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()
```

# 4 - Data Preprocessing

### 4.1 Separate Features and Target

```python
X = df[
    ["Area", "Bedrooms", "Bathrooms", "Age", "Location_Score"]
]

y = df["Price"]

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)
```

### 4.2 Check Multicollinearity using VIF

```python
!pip install statsmodels

import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Create VIF DataFrame
vif_data = pd.DataFrame()

# Store feature names
vif_data["Feature"] = X.columns

# Calculate VIF
vif_data["VIF"] = [
    variance_inflation_factor(X.values, i)
    for i in range(X.shape[1])
]

# Display results
print(vif_data.sort_values(by="VIF", ascending=False))
```

### 4.3 Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)

```

### 4.4 Verify Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Training data shape:", X_train_scaled.shape)
print("Testing data shape:", X_test_scaled.shape)


scaled_df = pd.DataFrame(
    X_train_scaled,
    columns=X.columns
)

print("Scaled Feature Means:")
print(scaled_df.mean().round(3))

print("\nScaled Feature Standard Deviations:")
print(scaled_df.std().round(3))
```

# 5 - Multiple Linear Regression from Scratch

### 5.1 Initialize Model Parameters

```python
n_samples, n_features = X_train_scaled.shape

weights = np.zeros(n_features)
bias = 0.0

learning_rate = 0.01
epochs = 5000

loss_history = []

print("Number of Features:", n_features)
print("Initial Weights:", weights)
print("Initial Bias:", bias)
```

### 5.2 Gradient Descent Implementation

```python
for epoch in range(epochs):

    # Forward propagation
    y_pred = np.dot(X_train_scaled, weights) + bias

    # Calculate errors
    errors = y_pred - y_train.values

    # Calculate gradients
    dw = (2 / n_samples) * np.dot(X_train_scaled.T, errors)

    db = (2 / n_samples) * np.sum(errors)

    # Update parameters
    weights -= learning_rate * dw
    bias -= learning_rate * db

    # Calculate loss
    loss = np.mean(errors ** 2)

    loss_history.append(loss)

print("Training Completed!")
print("Weights:", weights)
print("Bias:", bias)
print("Final Loss:", loss_history[-1])
```

### 5.3 Visualize Training Loss

```python
plt.figure(figsize=(8, 5))

plt.plot(loss_history)

plt.title("Multiple Linear Regression - Gradient Descent")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")

plt.grid(True)
plt.show()
```

# 6 - Prediction Using NumPy Model

```python
y_pred_scratch = np.dot(X_test_scaled, weights) + bias

print("Actual Prices:", y_test.values[:10])
print("Predicted Prices:", y_pred_scratch[:10])
```

```python
prediction_df = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred_scratch
})

prediction_df["Difference"] = (
    prediction_df["Actual Price"] -
    prediction_df["Predicted Price"]
)

prediction_df.head(10)
```

# 7 - Scikit-learn Implementation

### 7.1 Train Model

```python
sklearn_model = LinearRegression()

sklearn_model.fit(X_train, y_train)

print("Model Training Completed!")
print("Coefficients:", sklearn_model.coef_)
print("Intercept:", sklearn_model.intercept_)
```

### 7.2 Make Predictions

```python
y_pred_sklearn = sklearn_model.predict(X_test)

print("Actual Prices:", y_test.values[:10])
print("Predicted Prices:", y_pred_sklearn[:10])
```

### 7.3 Understand Model Coefficients

```python
coefficient_df = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": sklearn_model.coef_
})

coefficient_df
```

# 8 - Model Evaluation

### 8.1 Evaluation Function

```python
def evaluate_model(y_true, y_pred):

    mae = mean_absolute_error(y_true, y_pred)

    mse = mean_squared_error(y_true, y_pred)

    rmse = np.sqrt(mse)

    r2 = r2_score(y_true, y_pred)

    return {
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    }
```

### 8.2 Evaluate NumPy Model

```python
scratch_results = evaluate_model(
    y_test,
    y_pred_scratch
)

scratch_results
```

### 8.3 Evaluate Scikit-learn Model

```python
sklearn_results = evaluate_model(
    y_test,
    y_pred_sklearn
)

sklearn_results
```

### 8.4 Compare Both Models

```python
comparison_df = pd.DataFrame(
    [scratch_results, sklearn_results],
    index=["NumPy From Scratch", "Scikit-learn"]
)

comparison_df
```

# 10 - Custom House Price Prediction

### 10.1 Create Prediction Function

```python
def predict_house_price(
    area,
    bedrooms,
    bathrooms,
    age,
    location_score
):

    input_data = pd.DataFrame(
        [[area, bedrooms, bathrooms, age, location_score]],
        columns=X.columns
    )

    prediction = sklearn_model.predict(input_data)

    return prediction[0]
```

### 10.2 Test Custom Prediction

```python
predicted_price = predict_house_price(
    area=1800,
    bedrooms=3,
    bathrooms=2,
    age=5,
    location_score=8
)

print(f"Predicted House Price: {predicted_price:.2f} Lakh BDT")
```
