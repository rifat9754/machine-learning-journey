# K-Mean Clustering

# 1 - Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

# 2 - Load Data Set

```python
iris = load_iris()
X = iris.data[:,:2]
```

```python
X
```

# 3 - Data Visualization

```python
feature_names = iris.feature_names[:2]
X_df = pd.DataFrame(X, columns=feature_names)

plt.figure(figsize=(8, 5))
sns.histplot(X_df['sepal length (cm)'], kde=True, color='teal')

plt.title("Distribution of Sepal Length (Iris Dataset)")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```

# 5 - Model Training

```python
kmeans = KMeans(n_clusters=3,random_state=42)
kmeans.fit(X)
```

# 6 - Predictions

```python
y_kmeans = kmeans.predict(X)
centers = kmeans.cluster_centers_
```

# 7 - Visualization 

```python
plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, cmap='viridis') # Data points
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, marker='X') # Centroids
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('Simple K-Means Clustering on Iris Data')
plt.show()
```
